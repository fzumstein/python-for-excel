## Classes and Objects


```python
from dataclasses import dataclass
```

```python
@dataclass
class Car:
    # Data classes require type hints for attributes
    color: str
    speed: int = 0

    def accelerate(self, speed_increase):
        self.speed += speed_increase
```

```python
# Let's instantiate two car objects
car1 = Car("red")
car2 = Car(color="blue")
```

```python
# Attributes give you access to the data of an object
car1.color
```

```python
car1.speed
```

```python
# Calling the accelerate method on car1
car1.accelerate(20)
```

```python
# The speed attribute of car1 changed
car1.speed
```

```python
# The speed attribute of car2 remained the same
car2.speed
```

```python
car1.color = "green"
```

```python
car1.color
```

```python
car2.color  # unchanged
```

## Working with time-zone-aware datetime objects


```python
import datetime as dt
from zoneinfo import ZoneInfo
```

```python
# Time-zone-naive datetime object
timestamp = dt.datetime(2020, 1, 31, 14, 30)
timestamp.isoformat()
```

```python
# Time-zone-aware datetime object
timestamp_eastern = dt.datetime(
    2020, 1, 31, 14, 30, tzinfo=ZoneInfo("US/Eastern")
)
timestamp_eastern
```

```python
# Printing in isoformat makes it easy to
# see the offset from UTC
timestamp_eastern.isoformat()
```

```python
# Assign a time zone to a naive datetime object
timestamp_eastern = timestamp.replace(tzinfo=ZoneInfo("US/Eastern"))
timestamp_eastern.isoformat()
```

```python
# Convert from one time zone to another
timestamp_utc = timestamp_eastern.astimezone(ZoneInfo("UTC"))
timestamp_utc.isoformat()
```

```python
# From time-zone-aware to naive
timestamp_eastern.replace(tzinfo=None)
```

```python
# Current time without time zone
dt.datetime.now()
```

```python
# Current time in UTC time zone
dt.datetime.now(ZoneInfo("UTC"))
```

## Mutable vs. Immutable Objects


```python
a = [1, 2, 3]
b = a
a[1] = 22
print(a)
print(b)
```

```python
a = [1, 2, 3]
b = a.copy()
```

```python
a
```

```python
b
```

```python
a[1] = 22  # Changing "a"...
```

```python
a
```

```python
b  # ...doesn't affect "b"
```

```python
import copy

b = copy.deepcopy(a)
```

```python
def increment(x):
    x = x + 1
    return x
```

```python
a = 1
print(increment(a))
print(a)
```

```python
def increment(x):
    x[0] = x[0] + 1
    return x
```

```python
a = [1]
print(increment(a))
print(a)
```
